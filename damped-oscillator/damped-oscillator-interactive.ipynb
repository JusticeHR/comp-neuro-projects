Plan: 
Define the spring system.
Pick starting conditions.
Calculate acceleration.
Use acceleration to update velocity.
Use velocity to update position.
Repeat that many times.
Plot them.

```python
import matplotlib.pyplot as plt
from ipywidgets import interact

def simulate_spring(P, damping, dt, total_time):
    position = 1.0
    velocity = 0.0
    time = 0
    times = []
    positions = []

    while time < total_time:
        acceleration = -P * position - damping * velocity
        velocity = velocity + acceleration * dt
        position = position + velocity * dt
        times.append(time)
        positions.append(position)
        time = time + dt
    return times, positions

@interact(damping=(0.0, 2.0, 0.05))

def plot_springs(damping):
    # Frictionless spring
    times, frictionless = simulate_spring(
    P=1.0,
    damping=0,
    dt=0.01,
    total_time=10
    )
    
    # Damped spring using slider value
    times, damped = simulate_spring(
    P=1.0,
    damping=damping,
    dt=0.01,
    total_time=10
    )

    # Plot both
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    
    axes[0].plot(times, frictionless)
    axes[0].set_title("Frictionless Spring")
    axes[0].set_xlabel("Time")
    axes[0].set_ylabel("Position")
    axes[0].grid(True)
    
    axes[1].plot(times, damped)
    axes[1].set_title(f"Damped Spring (damping={damping:.2f})")
    axes[1].set_xlabel("Time")
    axes[1].set_ylabel("Position")
    axes[1].grid(True)
    
    plt.tight_layout()
    plt.show()
```
